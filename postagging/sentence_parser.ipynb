```python
# add more sentences from the file to the list
sentenceList = []
f = open('textAllFil_postagged_FINAL.txt', 'r')
lines = f.readlines()

gettingQuote = False
temp_sentence = ""

for line in lines:
    if line == '"|PMS':
        gettingQuote = not gettingQuote
    else:
        if gettingQuote:
            temp_sentence+=f'{line} '
        else:
            if len(temp_sentence) > 0:
                sentenceList.append(temp_sentence)
                temp_sentence = ""
            else:
                sentenceList.append(line)
f.close()
```

```python
import pandas as pd


sentenceDF = set(sentenceList)
sentenceDF = pd.DataFrame(sentenceDF,columns=['sentence'])
sentenceDF['sentence_without_postag'] = sentenceDF['sentence'].apply(lambda x: [a.split('|')[0] for a in x.split(' ')])
sentenceDF['sentence_without_postag'] = sentenceDF['sentence_without_postag'].apply(lambda x: ' '.join(x))
sentenceDF
```
