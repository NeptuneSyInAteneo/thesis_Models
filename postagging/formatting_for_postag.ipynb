```python
# the imports
import pandas as pd
import os
import re
```

```python
fileCount = 1
for file in os.listdir("../textData1_pymupdf"):
    fromFile = open(f"../textData1_pymupdf/{file}", 'r')
    formatted_text = fromFile.read()
    fromFile.close()

    formatted_text = re.sub(r'<[^\>]*>','',formatted_text) #get rid of the html tags
    formatted_text = re.sub(r'\n',' ',formatted_text) #turn endlines into spaces

    f = open(f"./no_tag/children_story_{fileCount}.txt", 'w')
    if "<span" in formatted_text: print(True, fileCount)
    f.write(formatted_text)
    f.close()

    fileCount+=1
```

```python
# use os to automate running the POS Tagger
#print(os.listdir('./no_tag'))
story_num = 1
# go to fspost for this to work

a = os.getcwd()
print(a)
os.chdir('../fspost')
print(os.getcwd())
for i in range(1,len(os.listdir('./no_tag'))+1):
    f = open(f"./no_tag/children_story_{story_num}.txt", 'r')
    command = f'java -mx4096m -cp "stanford-postagger.jar;" edu.stanford.nlp.tagger.maxent.MaxentTagger -model models/filipino-left5words-owlqn2-distsim-pref6-inf2.tagger -textFile ../../thesis_Models/postagging/no_tag/children_story_{i}.txt  > ../../thesis_Models/postagging/postagged/children_story_{story_num}_postagged.txt'
    os.system(command)
    f.close()
    print(f'story {story_num} has been postagged')
    story_num+=1
# back to current directory as to not break anything
os.chdir(a)
```

```python
#consolidate all into one txt file
write_this = ""
for i in os.listdir('./postagged'):
    f = open(f"./postagged/{i}", 'r')
    if ("<span" in f.read()) : print(True, i)
    write_this += f.read()
    f.close()

with open('./textAllFil_postagged.txt', 'w') as postagged_write:
    postagged_write.write(write_this)
```

```python
# format the postag file
# organize it properly
import re

tagged_Text = open("./textAllFil_postagged.txt", 'r', encoding="utf-8")

allText = tagged_Text.read()
#print(allText)
allText = re.sub(r'\|PMP \"\|PMS\n', '|PMP  \n\"|PMS ', allText)
allText = re.sub(r'\|PMQ \"\|PMS\n', '|PMQ  \n\"|PMS ', allText)
allText = re.sub(r'\|PME \"\|PMS\n', '|PME  \n\"|PMS ', allText)
allText = re.sub(r'\|PMS \"\|PMS\n', '|PMS  \n\"|PMS ', allText)
allText = re.sub(r'\|PMC \"\|PMS\n', '|PMC  \n\"|PMS ', allText)
allText = re.sub(r'\|PMSC \"\|PMS\n','|PMSC \n\"|PMS ', allText)

final_Text = open("./textAllFil_postagged_FINAL.txt", 'w')
final_Text.write(allText)
```
