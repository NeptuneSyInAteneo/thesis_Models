```python
# the imports
import pandas as pd
import os
import re

os.chdir("../../thesis_Models")
print(os.getcwd())
```

```python
fileCount = 1
for file in os.listdir("./textData1_pymupdf"):
    fromFile = open(f"./textData1_pymupdf/{file}", 'r')
    formatted_text = fromFile.read()
    fromFile.close()

    formatted_text = re.sub(r'<[^\>]*>','',formatted_text) #get rid of the html tags
    formatted_text = re.sub(r'\n',' ',formatted_text) #turn endlines into spaces
    
    separate_dot = list(set(re.findall(r'\.[A-Z]',formatted_text))) # list of every leftter that immediately succeeds a dot
    separate_question_mark = list(set(re.findall(r'\?[A-Z]',formatted_text))) 
    separate_exclamation_mark = list(set(re.findall(r'\![A-Z]',formatted_text))) 
    separate_comma = list(set(re.findall(r'\,[A-Z]',formatted_text))) 

    for i in range(len(separate_dot)):
        a = re.sub(r'\.','. ',separate_dot[i])
        formatted_text = formatted_text.replace(separate_dot[i], a)
    
    for i in range(len(separate_question_mark)):
        a = re.sub(r'\?','? ',separate_question_mark[i])
        formatted_text = formatted_text.replace(separate_question_mark[i], a)
    
    for i in range(len(separate_exclamation_mark)):        
        a = re.sub(r'\!','! ',separate_exclamation_mark[i])
        formatted_text = formatted_text.replace(separate_exclamation_mark[i], a)
    
    for i in range(len(separate_comma)):
        a = re.sub(r'\,',' ,',separate_comma[i])
        formatted_text = formatted_text.replace(separate_comma[i], a)

    f = open(f"./postagging/no_tag/children_story_{fileCount}.txt", 'w')
    f.write(formatted_text)
    f.close()

    fileCount+=1
```

```python
# use os to automate running the POS Tagger
#print(os.listdir('./no_tag'))
story_num = 1
# go to fspost for this to work

a = os.getcwd()
print(a)
os.chdir('./fspost')
print(os.getcwd())
for i in range(1,len(os.listdir('../../thesis_Models/postagging/no_tag'))+1):
    f = open(f"../../thesis_Models/postagging/no_tag/children_story_{story_num}.txt", 'r')
    command = f'java -mx2048m -cp "stanford-postagger.jar;" edu.stanford.nlp.tagger.maxent.MaxentTagger -model models/filipino-left5words-owlqn2-distsim-pref6-inf2.tagger -textFile ../../thesis_Models/postagging/no_tag/children_story_{i}.txt  > ../../thesis_Models/postagging/postagged/children_story_{story_num}_postagged.txt'
    os.system(command)
    f.close()
    print(f'story {story_num} has been postagged')
    story_num+=1
# back to current directory as to not break anything
os.chdir(a)
print("done")
print(os.getcwd())
```
