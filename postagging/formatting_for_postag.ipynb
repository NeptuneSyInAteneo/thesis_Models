```python
# the imports
import pandas as pd
import os
import re

os.chdir("../../thesis_Models")
print(os.getcwd())
```

```python
fileCount = 1
for file in os.listdir("./textData1_pymupdf"):
    fromFile = open(f"./textData1_pymupdf/{file}", 'r')
    formatted_text = fromFile.read()
    fromFile.close()

    formatted_text = re.sub(r'<[^\>]*>','',formatted_text) #get rid of the html tags
    formatted_text = re.sub(r'\n',' ',formatted_text) #turn endlines into spaces
    
    separate_dot = list(set(re.findall(r'\.[A-Z]',formatted_text))) # list of every word that are next to each other

    for i in range(len(separate_dot)):
        a = re.sub(r'\.','. ',separate_dot[i])
        print(a)
        formatted_text = formatted_text.replace(separate_dot[i], a)


    f = open(f"./postagging/no_tag/children_story_{fileCount}.txt", 'w')
    f.write(formatted_text)
    f.close()

    fileCount+=1
```

```python
# use os to automate running the POS Tagger
#print(os.listdir('./no_tag'))
story_num = 1
# go to fspost for this to work

a = os.getcwd()
print(a)
os.chdir('./fspost')
print(os.getcwd())
for i in range(1,len(os.listdir('../../thesis_Models/postagging/no_tag'))+1):
    f = open(f"../../thesis_Models/postagging/no_tag/children_story_{story_num}.txt", 'r')
    command = f'java -mx2048m -cp "stanford-postagger.jar;" edu.stanford.nlp.tagger.maxent.MaxentTagger -model models/filipino-left5words-owlqn2-distsim-pref6-inf2.tagger -textFile ../../thesis_Models/postagging/no_tag/children_story_{i}.txt  > ../../thesis_Models/postagging/postagged/children_story_{story_num}_postagged.txt'
    os.system(command)
    f.close()
    print(f'story {story_num} has been postagged')
    story_num+=1
# back to current directory as to not break anything
os.chdir(a)
print("done")
print(os.getcwd())
```

```python
#consolidate all into one txt file
print(os.getcwd())
write_this = ""
for i in os.listdir('./postagging/postagged'):
    print(i)
    f = open(f"./postagging/postagged/{i}", 'r')
    write_this += f.read()
    f.close()

with open('./postagging/textAllFil_postagged.txt', 'w') as postagged_write:
    postagged_write.write(write_this)
```

```python
# format the postag file
# organize it properly
import re

tagged_Text = open("./postagging/textAllFil_postagged.txt", 'r', encoding="utf-8")

allText = tagged_Text.read()
#print(allText)
allText = re.sub(r'\|PMP \"\|PMS\n', '|PMP  \n\"|PMS ', allText)
allText = re.sub(r'\|PMQ \"\|PMS\n', '|PMQ  \n\"|PMS ', allText)
allText = re.sub(r'\|PME \"\|PMS\n', '|PME  \n\"|PMS ', allText)
allText = re.sub(r'\|PMS \"\|PMS\n', '|PMS  \n\"|PMS ', allText)
allText = re.sub(r'\|PMC \"\|PMS\n', '|PMC  \n\"|PMS ', allText)
allText = re.sub(r'\|PMSC \"\|PMS\n','|PMSC \n\"|PMS ', allText)

final_Text = open("./postagging/textAllFil_postagged_FINAL.txt", 'w')
final_Text.write(allText)
```
