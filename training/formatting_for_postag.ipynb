```python
# the imports
import pandas as pd

```

```python
with open('textAllFil.txt') as all_stories:
    qq = all_stories.readlines()
    i = 0
    while i < len(qq):
        if qq[i] == "\n":
            qq.pop(i)
        else:
            i+=1
    print(len(qq))
```

```python
# make a text file for each
story_num = 1
for story in qq:
    f = open(f"children_story_{story_num}.txt",'w')
    f.write(story)
    f.close()
    story_num+=1
```
