```python
import pandas as pd

noun_categ = pd.read_json("./Organized_Text_data/nouns_categorized_assessed.json").drop(columns=["index"])
kw_dict_final = noun_categ[['tokens', 'just_category']].explode('just_category').groupby('just_category').agg(set).to_dict()['tokens']
kw_dict_final
```

```python
def get_categ_tag(token):
  #print(token)
  categ = []
  lemma = token.split('|')[0]
  postag = token.split('|')[1]

  for key in kw_dict_final:
    if token in kw_dict_final[key]: # this is a set
      categ.append(key)

  if len(categ) == 0:
    return "no_tag"
  elif len(categ) == 1 and ("NN" in postag):
    return categ[0] # the ideal option
  elif len(categ) == 1 and ("FW" in postag):
    return f"{categ[0]} #FOREIGNCATEG"
  elif len(categ) == 1 and not ("NN" in postag or "FW" in postag):
    return f"{categ[0]} #FALSEPOSITIVE"
  elif len(categ) > 1 and ("NN" in postag or "FW" in postag):
    return f"{str(categ)} #NEEDS MANUAL INTERVENTION"



def dataset_format_tag(sentence_list, file_name): # this will need some manual instervention later
  ds_file = open(file_name, 'w')
  for sentence in sentence_list:
    only_sentence = sentence
    # remove linebreaks
    only_sentence = only_sentence.replace('\n', '')

    # remove double space
    only_sentence = only_sentence.replace(r'  ', ' ')

    # remove trailing whitespace
    only_sentence = only_sentence.strip()

    #print(only_sentence)
    for token in only_sentence.split(' '):
      ds_file.write(f"{token}\t{get_categ_tag(token)}\n")
      pass
    ds_file.write('\n')
  ds_file.close()
```
